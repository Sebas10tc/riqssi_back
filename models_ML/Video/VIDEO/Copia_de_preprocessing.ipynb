```
import os
import glob
import cv2
import shutil
import numpy as np

# 1. Definir rutas
#ruta_origen = '/content/drive/MyDrive/prueba_video/'
ruta_origen = '/content/drive/MyDrive/FINAL_DATASET/ia/nuevos/'
ruta_destino = '/content/drive/MyDrive/FINAL_DATASET/ia/nuevos-filtrados/'

#'/content/drive/MyDrive/PROY_TESIS/REPOSITORIO_DATASET/test_ia_dataset/videos_filtro/'
# 2. Crear la carpeta de destino si no existe
if not os.path.exists(ruta_destino):
    os.makedirs(ruta_destino)
    print(f"Carpeta creada: {ruta_destino}")
else:
    print(f"La carpeta ya existe: {ruta_destino}")

# 3. Obtener lista de videos
video_files = glob.glob(os.path.join(ruta_origen, '*.mp4'))
videos_seleccionados = []
frame_counts = []

print(f"Iniciando análisis de {len(video_files)} videos...")

# 4. Filtrar y Copiar
for video_path in video_files:
    cap = cv2.VideoCapture(video_path)
    count = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
    cap.release()

    nombre_video = os.path.basename(video_path)

    if count >= 150:
        # Definir la ruta final del archivo
        destino_final = os.path.join(ruta_destino, nombre_video)

        # Copiar el archivo físicamente
        shutil.copy(video_path, destino_final)

        videos_seleccionados.append(destino_final)
        frame_counts.append(count)
        print(f"✅ Aceptado y Copiado: {nombre_video} ({count} frames)")
    else:
        print(f"❌ Descartado: {nombre_video} (Solo {count} frames)")

# 5. Resumen final
print("\n--- RESUMEN ---")
print(f"Total videos procesados: {len(video_files)}")
print(f"Videos guardados en nueva carpeta: {len(videos_seleccionados)}")
if frame_counts:
    print(f"Promedio de frames de los seleccionados: {np.mean(frame_counts):.2f}")
```

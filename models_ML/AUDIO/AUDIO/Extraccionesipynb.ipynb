```python
import os
import subprocess
import zipfile

input_folder = "/content/drive/MyDrive/PROY_TESIS/REPOSITORIO_DATASET/test_ia_dataset/videos_test"


# 📂 Carpetas de salida
audio_folder = "/content/drive/MyDrive/PROY_TESIS/REPOSITORIO_DATASET/test_ia_dataset/audio_extraido"
video_folder = "/content/drive/MyDrive/PROY_TESIS/REPOSITORIO_DATASET/test_ia_dataset/video_extraido"

# 📦 Archivos zip
audio_zip = "/content/drive/MyDrive/PROY_TESIS/REPOSITORIO_DATASET/test_ia_dataset/audios.zip"
video_zip = "/content/drive/MyDrive/PROY_TESIS/REPOSITORIO_DATASET/test_ia_dataset/videos.zip"
```

```python
# 🔁 Procesar todos los archivos .mp4
for file in os.listdir(input_folder):
    if file.endswith(".mp4"):
        input_path = os.path.join(input_folder, file)

        base_name = os.path.splitext(file)[0]
        audio_output = os.path.join(audio_folder, base_name + ".mp3")
        video_output = os.path.join(video_folder, base_name + ".mp4")

        print(f"Procesando: {file}")

        # 🎵 Extraer audio
        subprocess.run([
            "ffmpeg", "-i", input_path,
            "-ac", "1", "-ar", "16000",
            audio_output
        ])

        # 🎥 Extraer video sin audio
        subprocess.run([
            "ffmpeg", "-i", input_path,
            "-an", "-vcodec", "copy",
            video_output
        ])


print("✅ Pistas extraidas correctamente")
```

```python

# 📦 Función para zippear carpeta
def zip_folder(folder_path, zip_path):
    with zipfile.ZipFile(zip_path, 'w') as zipf:
        for root, dirs, files in os.walk(folder_path):
            for file in files:
                full_path = os.path.join(root, file)
                arcname = os.path.relpath(full_path, folder_path)
                zipf.write(full_path, arcname)

# 📦 Crear ZIPs
zip_folder(audio_folder, audio_zip)
zip_folder(video_folder, video_zip)

print("✅ Todo procesado y comprimido correctamente")
```

```python
import os
from moviepy.editor import VideoFileClip

# Carpeta donde están los videos
input_folder = "/content/drive/MyDrive/FINAL_DATASET/real/filtrados/"

# Carpeta donde se guardarán los audios
output_folder = "/content/drive/MyDrive/FINAL_DATASET/AUDIO/REAL_NUEVO/"

# Crear carpeta de salida si no existe
os.makedirs(output_folder, exist_ok=True)

# Extensiones de video permitidas
video_extensions = (".mp4", ".avi", ".mov", ".mkv")

# Recorrer videos
for filename in os.listdir(input_folder):

    if filename.lower().endswith(video_extensions):

        video_path = os.path.join(input_folder, filename)

        # Nombre del audio .wav
        audio_name = os.path.splitext(filename)[0] + ".wav"
        audio_path = os.path.join(output_folder, audio_name)

        try:
            # Cargar video
            video = VideoFileClip(video_path)

            # Extraer audio
            audio = video.audio

            # Guardar audio en WAV
            audio.write_audiofile(audio_path, codec='pcm_s16le')

            # Cerrar recursos
            audio.close()
            video.close()

            print(f"Audio extraído: {audio_name}")

        except Exception as e:
            print(f"Error con {filename}: {e}")

print("Proceso completado.")
```
